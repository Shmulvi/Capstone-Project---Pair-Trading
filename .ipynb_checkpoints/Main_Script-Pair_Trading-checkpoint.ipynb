```python
def pairing(dataframe):

    pairings = []
    # Removing the daily returns so it's not included in the pairings
    returns = dataframe.iloc[:,43:53]
    
    dataframe = dataframe.drop(columns=dataframe.iloc[:,43:53].columns.tolist())
    
    # Keep first 4 date columns
    date_df = dataframe.iloc[:, :3]

    # Remove date columns
    feature_df = dataframe.iloc[:, 3:]

    # Split remaining columns into groups of 10
    groups = [
        feature_df.iloc[:, i:i+5]
        for i in range(0, feature_df.shape[1], 5)
    ]

    # Generate every possible pair of groups
    for i, j in combinations(range(len(groups)), 2):

        group1 = groups[i]
        group2 = groups[j]

        # Get the corresponding target/return for each group
        target1 = returns.iloc[:, i]
        target2 = returns.iloc[:, j]

        # Combine dates, predictors and relevant targets
        pair_df = pd.concat(
            [date_df, group1, group2, target1, target2],
            axis=1
        )

        pairings.append(pair_df)
    
    return pairings
```

# Phase 3

I am supposed to be looking at the log of the prices in both directions (stock A then stock B) to view the relationship between the two stocks. After all the returns are just the indicator and the sclae you could earn, but this is based on the actual price movements

```python
def ols_loop(pairing_dict, pairing_list):

    models = {}
    
    for i in pairing_list:
        pair =  pairing_dict[i]   

        stock_A_close = pair.iloc[:,3]
        stock_B_close = pair.iloc[:,8]

        X = np.log(stock_A_close)
        y = np.log(stock_B_close)

        X = sm.add_constant(X)
        
        model_forward = sm.OLS(y, X).fit()

        X_back = np.log(stock_B_close)
        Y_back = np.log(stock_A_close)

        X_back = sm.add_constant(X_back)

        model_backward = sm.OLS(Y_back, X_back).fit()

        models[i] = {
            "forward": model_forward,
            "backward": model_backward
        }
        """
        print(f"======= {i} =======\n")
        print(model_forward.summary())
        print(model_backward.summary())
        """

    return models
```

```python
stat_sig = pd.DataFrame(index=Pairing_names,columns=['ADF_Statistic','EG_Test','Johansen_Test_trace','Johansen_Test_cvt'])

def stats_significane(dataframe,master_dict,resid,three):
    ADF = pd.DataFrame()
    EG = pd.DataFrame()
    Johansen = pd.DataFrame()
    for l in dataframe.index:
        adf_stat = adfuller(resid[l])
        adf_stat = pd.Series(data=adf_stat[0])
        ADF = pd.concat([ADF, adf_stat])

        if l not in three:
            eg_stat = coint(master_dict[l][f'Close-'+l[8:12]],master_dict[l][f'Close-'+l[13:17]])
            joh = coint_johansen(master_dict[l][[f'Close-'+l[8:12],f'Close-'+l[13:17]]], det_order = 0, k_ar_diff = 0)
            eg_stat = pd.Series(eg_stat[1])
            joh = pd.DataFrame({"Trace": [joh.lr1[0]],"CVT": [joh.cvt[0,2]]})
            EG = pd.concat([EG,eg_stat])
            Johansen = pd.concat([Johansen,joh])
        else:
            eg_statv2 = coint(master_dict[l][f'Close-'+l[8:11]],master_dict[l][f'Close-'+l[12:16]])
            johv2 = coint_johansen(master_dict[l][[f'Close-'+l[8:11],f'Close-'+l[12:16]]], det_order = 0, k_ar_diff = 0)
            eg_statv2 = pd.Series(eg_statv2[1])
            johv2 = pd.DataFrame({"Trace": [johv2.lr1[0]],"CVT": [johv2.cvt[0,2]]})
            EG = pd.concat([EG,eg_statv2])
            Johansen = pd.concat([Johansen,johv2])
        
    ADF.index = Pairing_names
    EG.index = Pairing_names
    Johansen.index = Pairing_names
    dataframe['ADF_Statistic'] = ADF
    dataframe['EG_Test'] = EG
    dataframe['Johansen_Test_trace'] = Johansen.iloc[:,0]
    dataframe['Johansen_Test_cvt'] = Johansen.iloc[:,1]
    
    dataframe = dataframe.sort_values(by=['ADF_Statistic','EG_Test','Johansen_Test_trace','Johansen_Test_cvt'],ascending=True)
    
    return dataframe

stat_significance = stats_significane(stat_sig,pairing_dict,residuals,three=['pairing_IBM_AVGO','pairing_IBM_NVDA'])
```

The above code is using takes the fitted lin reg model and predicts the values for the Y data. We then compute the residuals using the actual value minus the predicted (remember that train is just used for the training of the model and that X and Y is basically the split of target and input variables)

```python
def mean_std_2025(coin_pairs,pairing,year_data):
    
    X_25 = np.log(coin_pairs[f'{pairing}'].loc[f'{year_data}-01-01':f'{year_data}-12-31'][coin_pairs[f'{pairing}'].columns[3]])
    Y_25 = np.log(coin_pairs[f'{pairing}'].loc[f'{year_data}-01-01':f'{year_data}-12-31'][coin_pairs[f'{pairing}'].columns[8]])

    X_25 = sm.add_constant(X_25)
    
    model = sm.OLS(Y_25,X_25).fit()

    predicted = model.predict(X_25)
    residuals_man = Y_25 - predicted
    residuals_func = model.resid
    
    
    return residuals_func.mean(), residuals_func.std()
```

```python
def unseen_residual(coin_pairs,pairing,year_start_date, year_end_date):
    
    X_25 = np.log(coin_pairs[f'{pairing}'].loc[f'{year_start_date}-01-01':f'{year_end_date}-12-31'][coin_pairs[f'{pairing}'].columns[3]])
    Y_25 = np.log(coin_pairs[f'{pairing}'].loc[f'{year_start_date}-01-01':f'{year_end_date}-12-31'][coin_pairs[f'{pairing}'].columns[8]])

    X_25 = sm.add_constant(X_25)
    
    model = sm.OLS(Y_25,X_25).fit()

    predicted = model.predict(X_25)
    residuals = Y_25 - predicted

    return residuals
```

Now I will use the Z-score to standardise these residuals and the Z-scores here assign a mean of 0 and standard deviation of 1 so we can plot essentially how far any spread is from the mean and we can begin to determine if this gap is consistent. I believe this will tie in with the mean-reverting behaviour. 
2 is acceptable and means the result is two stds above the mean

```python
pairings_means = [pair1_mean_resid, pair2_mean_resid, pair3_mean_resid, pair4_mean_resid, pair5_mean_resid]
pairings_stds = [pair1_std_resid, pair2_std_resid, pair3_std_resid, pair4_std_resid, pair5_std_resid]
```

```python
def residuals_zscores_features(main_dict, pairs, year_s, year_e, means, std):
    for i in range(len(pairs)):
        main_dict[pairs[i]]['Residuals'] = unseen_residual(main_dict,pairs[i],year_s, year_e)
        main_dict[pairs[i]]['Z-score'] = main_dict[pairs[i]]['Residuals'].apply(lambda x: (x - means[i])/std[i])

    return main_dict

pairing_dict = residuals_zscores_features(pairing_dict, cointegrated_pairs,'2020', '2025', pairings_means, pairings_stds)
pairing_dict_26 = residuals_zscores_features(pairing_dict_26, cointegrated_pairs,'2026', '2026', pairings_means, pairings_stds)
```

To comment on all the work I have done on Z-Scores: The OLS model simplfies getting the residuals (diff between real vs predicted) and uses the log of the closing prices as inputs. Basically, the linear regression is using the stock A to predict stock B given their cointegration and the logarithm allows us to pick up moves easier (sensitivity). We use the trained model (2025) to generate the mean and standard devaition of the residuals so that we do not input the testing mean and std and overfit the model to test data. We then used an unseen period to calculate the daily Residuals and then standardize these residuals using the training statistics. Z-scores standardise by fixing about the mean and having it on a scale relative to the mean (x - mean/ std) which represents the diff in mean and value expressed as a percentage of how fair any point deviates from the mean) If they sit between 1 and -1 then it sits within the distribution

We are looking to capitalise on the movements moving about the mean and returning to it so pairings that typically cross the line would help us dicate when to enter a trade

 Long Spread: Enter when Z-score &lt; −2 and exit when Z-score ≥ 0.
 Short Spread: Enter when Z-score &gt; +2 and exit when Z-score ≤ 0.

Above was my first attempt, but then then I realised I need a position vairable to hold my current position and append to so I could know what I currently am at. My next trading decision is dependent on how I have entered/exited on the previous row

```python
def trade_signals(df, threshold):

    position = None
    signals = []

    for z in df['Z-score']:

        if position is None and z < -(threshold):
            signal = 'Long Enter'
            position = 'Long'
    
        elif position == 'Long' and z >= 0:
            signal = 'Long Exit'
            position = None
    
        elif position is None and z > threshold:
            signal = 'Short Enter'
            position = 'Short'
    
        elif position == 'Short' and z <= 0:
            signal = 'Short Exit'
            position = None
    
        else:
            signal = 'Null'
    
        signals.append(signal)

    df['Signal'] = signals

    return df
```

```python
def time_series_features(pairs, coin_pairs):
    for x in coin_pairs:
        pairs[x]['lag1'] = pairs[x]['Z-score'].shift(1)
        pairs[x]['lag2'] = pairs[x]['Z-score'].shift(2)
        pairs[x]['lag3'] = pairs[x]['Z-score'].shift(3)
        pairs[x]['rolling_std_20'] = pairs[x]['Z-score'].rolling(window=20).std()

    return pairs

pairing_dict_26 = time_series_features(pairing_dict_26, cointegrated_pairs)
pairing_dict = time_series_features(pairing_dict, cointegrated_pairs)
```

```python
seq_len = 20  # number of past time steps (days) to use
forecast = 5

def create_sequences(X, y, seq_len, forecast):
    Xs, ys = [], []
    
    for i in range(len(X) - seq_len - forecast + 1):
        
        Xs.append(X[i:i+seq_len])
        ys.append(y[i+seq_len:i+seq_len+forecast])
    
    return np.array(Xs), np.array(ys)
```

```python
import torch
import torch.nn as nn

def prepare_training_data(pairing_name, pairing_dict, seq_len, forecast):

    features = ['Z-score', 'lag1', 'lag2', 'lag3', 'rolling_std_20']
    target = ['Z-score']

    X_data = pairing_dict[pairing_name][features].fillna(0)
    Y_data = pairing_dict[pairing_name][target]

    X_array = X_data.values
    Y_array = Y_data.values

    X_sequences, Y_sequences = create_sequences(X_array, Y_array, seq_len, forecast)

    X_tensor = torch.tensor(X_sequences, dtype=torch.float32)
    Y_tensor = torch.tensor(Y_sequences, dtype=torch.float32)

    return X_tensor, Y_tensor

training_tensors = {}

for pair in cointegrated_pairs:
    X_tensor, Y_tensor = prepare_training_data(pair, pairing_dict, seq_len, forecast)

    training_tensors[pair] = {
        'X': X_tensor,
        'Y': Y_tensor
    }
```

```python
def initialise_lstm():

    lstm = nn.LSTM(
        input_size=5,
        hidden_size=8,
        batch_first=True
    )

    fc = nn.Linear(8, 5)

    loss = nn.MSELoss()

    optimizer = torch.optim.Adam(
        list(lstm.parameters()) + list(fc.parameters()),
        lr=0.01
    )

    return lstm, fc, loss, optimizer
```

Through my research in the LSTM models I learned that for a contious number prediction (as is the case for a time series problem like this) the Sigmoid function is not needed, that is only needed for binary classification. I should still be using the Linear on the output but I do not need an activation layer for my problem

Since I am forecasting the next 5+ trading days I needed to add in the previous days so that we could use the 20 day lookback window in the sequence creation function as these training inputs are valid for the test starting data.

```python
from sklearn.metrics import mean_squared_error
from sklearn.metrics import mean_absolute_error


def test_evaluation(pairing_data):

    results = {}

    for pair_id, data in pairing_data.items():

        lstm = data["lstm"]
        fc = data["fc"]

        lstm.eval()
        fc.eval()

        with torch.no_grad():

            out, _ = lstm(data["X_test"])
            predictions = fc(out[:, -1, :])

            actual = data["Y_test"].squeeze(-1)

            test_loss = data["loss"](predictions, actual)

        predictions = predictions.cpu().numpy()
        actual = actual.cpu().numpy()

        horizon_metrics = {}

        for horizon in range(5):

            rmse = np.sqrt(
                mean_squared_error(
                    actual[:, horizon],
                    predictions[:, horizon]
                )
            )

            mae = mean_absolute_error(
                actual[:, horizon],
                predictions[:, horizon]
            )

            horizon_metrics[f"t+{horizon+1}"] = {
                "RMSE": rmse,
                "MAE": mae
            }

        results[pair_id] = {
            "pairing_name": data["name"],
            "predictions": predictions,
            "actual": actual,
            "MSE": test_loss.item(),
            "horizon_metrics": horizon_metrics
        }

    return results
```

```python
test_results = test_evaluation(pairing_test_data)
```

```python
def lstm_trade_signals(df, threshold):

    position = None
    signals = []

    for _, row in df.iterrows():

        z = row['Z-score']
        forecast = row['t+1']

        if position is None and z < -(threshold) and forecast > z:
            signal = 'Long Enter'
            position = 'Long'

        elif position == 'Long' and z >= 0:
            signal = 'Long Exit'
            position = None

        elif position is None and z > threshold and forecast < z:
            signal = 'Short Enter'
            position = 'Short'

        elif position == 'Short' and z <= 0:
            signal = 'Short Exit'
            position = None

        else:
            signal = 'Null'

        signals.append(signal)

    df = df.copy()
    df['Signal'] = signals

    return df
```

```python
def prepare_forecast_df(test_input_df, test_results, seq_len=20):

    forecast_data = {}
    actual_data = {}

    for pair_id, pair_results in test_results.items():

        pairing_name = pair_results['pairing_name']
    
        predictions = pair_results['predictions']
        actual = pair_results['actual']
        pair_test_input_df = test_input_dfs[pairing_name]

    
        origin_dates = pair_test_input_df.index[
            seq_len - 1 : seq_len - 1 + len(predictions)
        ]
    
        actual_dates = pair_test_input_df.index[
            seq_len : seq_len + len(actual)
        ]
    
        forecast_data[pairing_name] = pd.DataFrame(
            predictions,
            index=origin_dates,
            columns=['t+1', 't+2', 't+3', 't+4', 't+5']
        )
        
        forecast_data[pairing_name]['Z-score'] = (
            pair_test_input_df.loc[origin_dates, 'Z-score'].values
        )
    
        actual_data[pairing_name] = pd.DataFrame(
            actual,
            index=actual_dates,
            columns=['t+1', 't+2', 't+3', 't+4', 't+5']
        )

    return forecast_data, actual_data
```

```python
beta_dict = {}

for pairing_name, pair_df in pairing_dict_26.items():

    stock_a = pairing_name.replace('pairing_', '').split('_')[0]
    stock_b = pairing_name.replace('pairing_', '').split('_')[1]

    X = np.log(pair_df[f'Close-{stock_a}'])
    y = np.log(pair_df[f'Close-{stock_b}'])

    X = sm.add_constant(X)

    ols_model = sm.OLS(y, X).fit()

    beta_dict[pairing_name] = ols_model.params.iloc[1]
```

```python
def evaluate_strategy(signal_df, price_df, beta, stock_a, stock_b):

    df = signal_df.copy()

    # Daily hedged spread returns
    spread_return = (
        np.log(price_df[f'Close-{stock_b}']).diff()
        - beta * np.log(price_df[f'Close-{stock_a}']).diff()
    )

    position = 0
    positions = []

    for signal in df['Signal']:
        if signal == 'Long Enter':
            position = 1
        elif signal == 'Short Enter':
            position = -1
        elif signal in ['Long Exit', 'Short Exit']:
            position = 0

        positions.append(position)

    df['Position'] = positions
    # Apply positions from the previous day's signal
    df['Strategy Return'] = (
        df['Position'].shift(1) * spread_return
    ).fillna(0)

    # Identify completed trades
    entries = df['Signal'].isin(['Long Enter', 'Short Enter'])
    exits = df['Signal'].isin(['Long Exit', 'Short Exit'])

    trade_returns = []
    trade_return = 0
    in_trade = False

    for position, daily_return, entry, exit_ in zip(
        df['Position'], df['Strategy Return'], entries, exits
    ):
        if entry:
            trade_return = 0
            in_trade = True

        if in_trade:
            trade_return += daily_return

        if exit_ and in_trade:
            trade_returns.append(trade_return)
            in_trade = False

    total_trades = len(trade_returns)

    # Total strategy return
    total_return = (1 + df['Strategy Return']).prod() - 1

    # Percentage of completed trades that were profitable
    win_rate = (
        sum(r > 0 for r in trade_returns) / total_trades
        if total_trades > 0 else np.nan
    )

    return {
        'Trades': total_trades,
        'Total Return': total_return,
        'Win Rate': win_rate
    }
```

With all the strategies I have compiled are now saved as dicts I can run a comparison function to see which has an edge in between the three.

```python
def compare_strategies(*strategy_dicts):

    results = []

    for name, strategy in strategy_dicts:
        for pair, metrics in strategy.items():
            results.append({
                'Pair': pair,
                'Strategy': name,
                'Trades': metrics['Trades'],
                'Total Return': metrics['Total Return'],
                'Win Rate': metrics['Win Rate']
            })

    return pd.DataFrame(results).sort_values(
        ['Pair', 'Strategy']
    ).reset_index(drop=True)
```
